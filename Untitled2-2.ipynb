```python

import numpy as np
import matplotlib.pyplot as plt

initial_investment = 100000
years = 10
simulations = 1000

annual_return = 0.10
annual_risk = 0.15

results = []

for i in range(simulations):
    investment = initial_investment

    for year in range(years):
        yearly_return = np.random.normal(annual_return, annual_risk)
        investment = investment * (1 + yearly_return)

    results.append(investment)

average_value = np.mean(results)
best_value = np.max(results)
worst_value = np.min(results)

print("MONTE CARLO PORTFOLIO ANALYSIS")
print("Initial Investment: ₹", initial_investment)
print("Number of Simulations:", simulations)
print("Investment Period:", years, "years")
print("Average Final Value: ₹", round(average_value, 2))
print("Best Case Value: ₹", round(best_value, 2))
print("Worst Case Value: ₹", round(worst_value, 2))

plt.hist(results, bins=30, edgecolor="black")
plt.title("Monte Carlo Portfolio Simulation")
plt.xlabel("Final Portfolio Value (₹)")
plt.ylabel("Number of Simulations")
plt.show()
```
